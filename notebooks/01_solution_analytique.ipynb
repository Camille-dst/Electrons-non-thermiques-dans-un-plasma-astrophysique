on a que la formule analytique est : $\frac{\partial N(\gamma, t)}{\partial t} = - \frac{\partial}{\partial \gamma} \left[ \dot{\gamma}_{\text{cool}} N(\gamma, t) \right] + Q(\gamma) - \frac{N(\gamma, t)}{t_{\text{esc}}}$

on a que $\frac{\gamma}{\tau_{cool}(\gamma)}$ est le terme de refroidissement avec 
$\tau_{cool} = \frac{3 mc}{4(u_B - u_{soft})\cdot \sigma_T \gamma}$

dans un premier temps $u_{soft} =0$ et $u_B = \frac{B^2}{8 \pi}$ avec $B$ le champ magnétique en $G$ avec $G = 10^{-4} T$ 


On a aussi que  le terme d'injection $Q(\gamma) = Q_0 \gamma ^{-s}$ donc a une relation de proportionalité.
On peut considèrer que cette relation de proportionalité dépend de $\gamma_{max}$ choisis tel que $Q(\gamma) = Q_0 \gamma ^{-s} \cdot e^{-\frac{\gamma}{\gamma_{max}}}$

Pour finir ,$t_{esc} = \frac{R}{v_s}$ avec R la taille caractéritique de la source en $pc$ avec $pc = 3.086 10^{13}$ km  et $v_s \approx c$ la vitesse du plasma en écoulement.

Remarque : en CGS, L--> cm, M--> g, $\sigma_T --> cm^2$

Les constantes sont : $\sigma_T = 6.6524 \cdot 10^{-25} cm^2$, $m_e = 9.1094 \cdot 10^{-28} g $ et $c = 2.9979 \cdot 10^{10} cm/s$

\section*{Synthèse de l'Étape 1 : Modèle Physique et Solution Analytique Exacte}

\subsection*{1. Équation maîtresse et cadre physique}

Sous les hypothèses de cette première partie :
\begin{itemize}
    \item \textbf{Refroidissement Synchrotron ($u_{\text{soft}} = 0$)} : $\dot{\gamma}_{\text{cool}} = -\beta \gamma^2$ avec $\beta = \frac{4}{3} \frac{\sigma_T c}{m_e c^2} u_B$.
    \item \textbf{Injection en loi de puissance} : $Q(\gamma) = Q_0 \gamma^{-s}$ pour $\gamma_{\min} \le \gamma \le \gamma_{\max}$.
    \item \textbf{Échappement} : $t_{\text{esc}} = R / c$.
\end{itemize}

\subsection*{2. Résolution analytique de l'état stationnaire}
À l'état stationnaire ($\partial N / \partial t = 0$), la réécriture de l'équation avec la fonction auxiliaire $I(\gamma) = \beta \gamma^2 N(\gamma)$ donne :
\begin{equation}
\frac{\mathrm{d}Y(\gamma)}{\mathrm{d}\gamma} - \frac{\gamma_c}{\gamma^2} Y(\gamma) = -Q(\gamma) \quad \text{avec} \quad \gamma_c = \frac{1}{\beta t_{\text{esc}}}
\end{equation}

L'introduction du facteur intégrant $\mu(\gamma) = e^{\gamma_c / \gamma}$ et l'intégration entre $\gamma$ et $\gamma_{\max}$ (avec la condition $N(\gamma_{\max}) = 0$) conduisent à la solution exacte :
\begin{equation}
N_{\text{stat}}(\gamma) = \frac{1}{\beta \gamma^2} \, e^{-\frac{\gamma_c}{\gamma}} \int_{\gamma}^{\gamma_{\max}} Q(\gamma') \, e^{\frac{\gamma_c}{\gamma'}} \, \mathrm{d}\gamma'
\end{equation}

\subsection*{3. Implémentation numérique de référence}
La solution exacte a été implémentée en Python via l'intégration numérique par quadrature (\texttt{scipy.integrate.quad}). La courbe obtenue présente bien la cassure spectrale attendue au niveau du Lorentz factor critique $\gamma_c$. Cette solution sert de référence absolue (\textit{ground truth}) pour valider les futurs schémas numériques (explicite, implicite, Chang \& Cooper).
